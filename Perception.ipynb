```python
import pandas as pd
import numpy as np
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score
```

```python
# 1. GENERATING GROUND TRUTH DATA MATRICES
print("Step 1: Constructing logic gate training datasets...")
```

```python
# The 4 coordinate points representing binary pairs
X_logic = np.array([
    [0,0],
    [1,0],
    [0,1],
    [1,1]
])
```

```python
# Target outcomes for both logic problems
y_and = np.array([0,0,0,1])
y_or = np.array([0,1,1,1])
```

```python
# 2. TRAINING THE AND GATE PERCEPTRON
print("Step 2: Fitting linear Perceptron for AND logic matrix...")
```

```python
# max_iter restricts training loop cycles; tol defines convergence termination threshold
model_and = Perceptron(
    max_iter=100,
    tol=1e-3,
    random_state=1
    )
model_and.fit(
    X_logic,
    y_and
    )
```

```python
# 3. TRAINING THE OR GATE PERCEPTRON
print("Step 3: Fitting linear Perceptron for OR logic matrix...")
model_or = Perceptron(
    max_iter=100,
    tol=1e-3,
    random_state=42
    )
model_or.fit(
    X_logic,
    y_or
    )
```

```python
# 4. SYSTEM METRICS & PERFORMANCE MONITORING
print("\n=== PERCEPTRON MODEL LOGS ===")
# Generate predictions across the base training inputs
preds_and = model_and.predict(X_logic)
preds_or = model_or.predict(X_logic)
print(f"AND Gate Model COnvergence Accuracy: {accuracy_score(y_and, preds_and)*100:.1f}%")
print(f"AND Calculated Weights: {model_and.coef_[0]} | Bias Offset: {model_and.intercept_[0]}")
print("-" * 60)
print(f"OR Gate Model Convergence Accuracy : {accuracy_score(y_or, preds_or) * 100:.1f}%")
print(f"OR Calculated Weights : {model_or.coef_[0]} | Bias Offset: {model_or.intercept_[0]}")
```

```python
# 5. LIVE INFERENCE ENGINE
print("\n=== REAL-TIME NEURON INFERENCE SERVICE ===")
# Manually enter a list of raw signals simulating external system inputs
unseen_raw_signals = pd.DataFrame([
    {'Signal_A':1, 'Signal_B':0},
    {'Signal_A':1, 'Signal_B':1}
])
```

```python
# Convert DataFrame directly to a clean numpy array for processing
inference_matrix = unseen_raw_signals.values
```

```python
# Generate predictions using both models
live_and_verdicts = model_and.predict(inference_matrix)
live_or_verdicts = model_or.predict(inference_matrix)
```

```python
# Display structured results using a clean table dashboard (No loops or zip syntax!)
inference_dashboard = pd.DataFrame({
    'Input Signal Vector': ['[1,0]','[1,1]'],
    'AND Prediction': live_and_verdicts,
    'OR Prediction': live_or_verdicts
})
```

```python
# Map binary integers to clear functional descriptions
inference_dashboard['AND Decision']= inference_dashboard['AND Prediction'].map({0:'Deactivated',1:'Activated'})
inference_dashboard['OR Decision']= inference_dashboard['OR Prediction'].map({0:'Deactivated',1:'Activated'})
```
