```python
import pandas as pd
import numpy as np
```

```python
df_sales = pd.read_csv("/content/advanced_pandas_sales_dataset (1).csv")
```

```python
df_sales.info()
```

```python
print(df_sales.head())
```

```python
category_revenue=df_sales.groupby('Category')['Revenue'].sum().reset_index()
```

```python
print("---Total Revenue by Category---")
print(category_revenue)
```

```python
#Find total item sold (Sum) and average unit price (mean)
region_metrics = df_sales.groupby("Region").agg(
    Total_Unit_Sold=("Quantity","sum"),
    Average_Item_Price=("UnitPrice","mean")
).reset_index()
```

```python
print("\nAdvanced Metrics by Region")
print(region_metrics)
```

```python
managers_data={
    "Region":("East","West","North","South"),
    "Regional_Director":("Alice Cooper", "Bob Marley", "Charlie Chaplin", "David Backham"),
}
df_managers = pd.DataFrame(managers_data)
```

```python
#merge the main sales log with tge managers database using "Region" as the link key
df_enriched = pd.merge(df_sales,df_managers, on="Region", how="left")

print("---Enriched Dataset---")
print(df_enriched[["OrderID","Region","Category","Revenue","Regional_Director"]].head(3))
```

```python
revenue_pivot = pd.pivot_table(
    df_sales,
    values="Revenue",
    index="Region",
    columns="Category",
    aggfunc="sum"
)
print("--- Revenue Pivot Table(Region vs. Category) ---")
print(revenue_pivot)
```

```python
order_counts_crosstab = pd.crosstab(
    index = df_sales["Region"],
    columns = df_sales["Category"]
)
print("--- Order Frequency Crosstab (Transaction Counts) ---")
print(order_counts_crosstab)
```

```python
print(f"Original OrderDate column data type: {df_sales['OrderDate']}")
```

```python
df_sales["OrderDate"]=pd.to_datetime(df_sales["OrderDate"])
print(f"Converted OrderDate column data type: {df_sales["OrderDate"].dtype}")
```

```python
df_sales["Month_Name"]= df_sales["OrderDate"].dt.strftime("%B") # Full Month Name
df_sales["Day_of_Week"]= df_sales["OrderDate"].dt.day_name()
print(df_sales.head())
```

```python
monthly_trends=df_sales.groupby(df_sales["OrderDate"].dt.to_period("M"))["Revenue"].sum().reset_index()
```

```python
print("--- Monthly Revenue Trends ---")
print(monthly_trends)
```

```python
df_election = pd.read_csv("/content/election_data.csv")
df_district = pd.read_csv("/content/district_data (1).csv")
df_province = pd.read_csv("/content/province_data.csv")
```

```python
print("--- Initial Shapes ---")
print(f"Election Data: {df_election.shape}")
print(f"District Data: {df_district.shape}")
print(f"Province Data: {df_province.shape}")
```

```python
print("Election Sample key:\n", df_election[["candidate_name","district_id"]].head(2))
print("\nDistrict Sample key:\n", df_district[["district_name","district_id","province_id"]].head(2))
print("\nProvince Sample key:\n", df_province.head(2))
print("-"*60)
```

```python
df_candidate_district = pd.merge(
    left=df_election,
    right=df_district,
    on="district_id",
    how="inner"
)
print(df_candidate_district.shape)
print(df_candidate_district[["candidate_name","party","district_name","votes"]].head(3))
```

```python
df_distict_left = pd.merge(
    left = df_district,
    right = df_province,
    on = "province_id",
    how = "left"
)
print(df_distict_left.shape)
print(df_distict_left[["district_name","province_id","province_name"]].head(3))
```

```python
df_master_election= pd.merge(
    df_election,
    df_district,
    on="district_id",
    how="inner"
).merge(
    df_province,
    on="province_id",
    how="left"
)
print(df_master_election.shape)
```

```python
print(df_master_election[["candidate_name","district_id","district_name","province_id","province_name"]].head(5))
```
