```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv("/content/regression_training_dataset1_5537be90-411f-4a6b-8ea4-7bb197181da5_268917_.csv")
print("--- Dataset Initial Shape ---")
print(f"Total Rows: {df.shape[0]}, Total Columns: {df.shape[1]}")
```

```python
x = df.drop(columns = ["PerformanceScore"])
y = df["PerformanceScore"]
```

```python
y_stratified_bins = pd.qcut(y, q=5, labels=False)
```

```python
x_train_val, x_test, y_train_val, y_test = train_test_split(
    x, y,
    test_size=0.2,
    random_state=42,
    stratify=y_stratified_bins
    )
```

```python
y_train_val_bins = pd.qcut(y_train_val, q=5, labels=False)

x_train, x_val, y_train, y_val = train_test_split(
    x_train_val, y_train_val,
    test_size=0.25,
    random_state=42,
    stratify=y_train_val_bins
    )
```

```python
print(x_train.shape)
print(x_val.shape)
print(x_test.shape)
```

```python
print("\n--- Target Stratification Verification (Mean Target Value) ---")
print(f"{y.mean():.4f}")
print(f"{y_train.mean():.4f}")
print(f"{y_val.mean():.4f}")
print(f"{y_test.mean():.4f}")
```

```python
standard_scaler = StandardScaler()
```

```python
standard_scaler.fit(x_train)
```

```python
x_train_scaled_std = standard_scaler.transform(x_train)
x_val_scaled_std = standard_scaler.transform(x_val)
x_test_scaled_std = standard_scaler.transform(x_test)
```

```python
min_max_scaler = MinMaxScaler()
```

```python
min_max_scaler.fit(x_train)
```

```python
x_train_scaled_norm = min_max_scaler.transform(x_train)
x_val_scaled_norm = min_max_scaler.transform(x_val)
x_test_scaled_norm = min_max_scaler.transform(x_test)
```

```python
df_train_std = pd.DataFrame(x_train_scaled_std, columns=x_train.columns)
df_train_norm = pd.DataFrame(x_train_scaled_norm, columns=x_train.columns)
```

```python
print("\n--- Feature Scaling Properties Verification ---")
print("1. Raw Training Range ('AttendancePercent') : Min =",
x_train['AttendancePercent'].min(), " | Max =", x_train['AttendancePercent'].max())

print("2. Standardized Metrics ('AttendancePercent') : Mean =",
round(df_train_std['AttendancePercent'].mean(), 2), " | Std =",
round(df_train_std['AttendancePercent'].std(), 2))

print("3. Normalized Range ('AttendancePercent') : Min =",
df_train_norm['AttendancePercent'].min(), " | Max =",
df_train_norm['AttendancePercent'].max())
```
