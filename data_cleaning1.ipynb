```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/content/Finance_data.csv")
```

```python
print("---Data Summary---")
df.info()
```

```python
print("\n---Missing Value Tally---")
print(df.isnull().sum())
```

```python
# Remove rows with missing age
df = df.dropna(subset=['age'])

# Fill missing Investment_Avenues with the mode
df['Investment_Avenues'] = df['Investment_Avenues'].fillna(
    df['Investment_Avenues'].mode()[0]
)
```

```python
df['age'] = df['age'].replace({
    'Twenty-three': 23})

df['age'] = pd.to_numeric(df['age'], errors='coerce')
```

```python
df['age'].dtype
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(' ', '_')
      .str.replace('?', '', regex=False)
)
```

```python
df.rename(columns={
    'Stock_Marktet': 'Stock_Market'
}, inplace=True)
```

```python
cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:
    df[col] = df[col].astype(str).str.strip().str.title()
```
